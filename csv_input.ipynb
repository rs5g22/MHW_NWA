```python
import copernicusmarine
import xarray as xr
import pandas as pd

```

```python
# Define dataset and query parameters
data_request = {
    "METOFFICE-GLO-SST-L4-REP-OBS-SST": "METOFFICE-GLO-SST-L4-REP-OBS-SST",
    "longitude": [-77, -65],  # Longitude bounds
    "latitude": [34, 45],     # Latitude bounds
    "time": ["1990-01-01", "2019-12-31"],  # Time range
    "variables": ["analysed_sst"]  # Variable of interest
}

# Open dataset from Copernicus Marine
sst_l4 = copernicusmarine.open_dataset(
    dataset_id=data_request["METOFFICE-GLO-SST-L4-REP-OBS-SST"],
    minimum_longitude=data_request["longitude"][0],
    maximum_longitude=data_request["longitude"][1],
    minimum_latitude=data_request["latitude"][0],
    maximum_latitude=data_request["latitude"][1],
    start_datetime=data_request["time"][0],
    end_datetime=data_request["time"][1],
    variables=data_request["variables"]
)

# Convert to pandas DataFrame and export as CSV
sst_l4.to_dataframe().to_csv("climatology_30_year.csv")
```
